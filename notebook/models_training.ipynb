```python
from sklearn.model_selection import train_test_split
from model_testing import preprocessor
from imblearn.pipeline import Pipeline as imblearnPipeline
from imblearn.over_sampling import SMOTE
from sklearn.model_selection  import cross_val_score
import optuna
import optuna.logging
import warnings
```

```python
# hyper paramater tuning using bayesian optimization, xgboost classifier

warnings.filterwarnings('ignore')

optuna.logging.set_verbosity(optuna.logging.ERROR)

def objective(trail):

    x_train,x_test,y_train,y_test = train_test_split(x,y)

    preprocessing = preprocessor(numeric_features=numric_features,categorical_features=cat_features)


    params = {
        "n_estimators":trail.suggest_int("n_estimators",50,400,step = 50),
        "max_depth":trail.suggest_int("max_depth",3,7),
        "learning_rate":trail.suggest_float("learning_rate",0.01,0.2),
        "subsample":trail.suggest_float("subsample",0.6,0.9),
        "colsample_bytree":trail.suggest_float("colsample_bytree",0.6,0.9),
    }

    model = XGBClassifier(**params)

    training_pipeline = imblearnPipeline([
        ("preprocessor",preprocessing),
        ("smote",SMOTE(random_state=42)),
        ("model",model)
    ])

    score = cross_val_score(training_pipeline,x_train,y_train,cv = 5, scoring="accuracy").mean()

    return score

study = optuna.create_study(direction="maximize")

study.optimize(objective,n_trials=100)
```

```python
print(f"Best parameters, {study.best_params}")
print(f"Best value, {study.best_value}")
```

```python
best_params = study.best_params
```
